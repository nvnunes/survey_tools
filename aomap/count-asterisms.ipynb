### Imports

```python
from IPython.display import display, HTML
from astropy.table import Table
import matplotlib.pyplot as plt
import numpy as np
from tabulate import tabulate
```

### Options

```python
ao_system_name = 'GNAO'
fits_path = '../output/asterisms-2025-07'
```

### Load Asterisms

```python
asterisms = Table.read(f"{fits_path}/asterisms-{ao_system_name}.fits", format='fits')

is_three_star = asterisms['num_stars'] == 3
is_two_star = asterisms['num_stars'] == 2

print(f'Total Asterisms: {len(asterisms)}')
print(f'Total 3-star Asterisms: {np.sum(is_three_star)}')
print(f'Total 2-star Asterisms: {np.sum(is_two_star)}')
```

```python
column_info = np.array([(col, asterisms[col].dtype) for col in asterisms.colnames])
display(tabulate(column_info, headers=['Name', 'Type'], tablefmt='html'))
```

### Counts by R Magnitude Limit

```python
mag_start = 8.0
mag_end = 19.0
mag_step = 0.5

counts3 = Table(names=('mag_limit', 'count', 'change'), dtype=('f8', 'i4', 'i4'))
counts2 = Table(names=('mag_limit', 'count', 'change'), dtype=('f8', 'i4', 'i4'))
prev_count3 = 0
prev_count2 = 0
for mag in np.arange(mag_start, mag_end, mag_step):
    mask = is_three_star & (asterisms['star1_mag'] >= mag_start) & (asterisms['star1_mag'] < mag) & (asterisms['star2_mag'] >= mag_start) & (asterisms['star2_mag'] < mag) & (asterisms['star3_mag'] >= mag_start) & (asterisms['star3_mag'] <=mag)
    count = np.sum(mask)
    counts3.add_row((mag, count, count - prev_count3))
    prev_count3 = count

    mask = is_two_star & (asterisms['star1_mag'] >= mag_start) & (asterisms['star1_mag'] < mag) & (asterisms['star2_mag'] >= mag_start) & (asterisms['star2_mag'] < mag)
    count = np.sum(mask)
    counts2.add_row((mag, count, count - prev_count2))
    prev_count2 = count
```

```python
display(HTML("<h3>3-Star Asterism Counts by R Magnitude Limit</h3>"))
display(tabulate(counts3, headers=['R Limit', 'Count', 'Change'], floatfmt=['.1f', '.0f', '.0f'], tablefmt='html'))

display(HTML("<h3>2-Star Asterism Counts by R Magnitude Limit</h3>"))
display(tabulate(counts2, headers=['R Limit', 'Count', 'Change'], floatfmt=['.1f', '.0f', '.0f'], tablefmt='html'))
```

```python
def plot_mag_comparison(asterisms, method, nominal_mag, dim_mag):
    if method == 'mean':
        mean_mag = np.mean([asterisms['star1_mag'], asterisms['star2_mag'], asterisms['star3_mag']], axis=0)
        is_three_star_bright = is_three_star & (mean_mag < nominal_mag)
        is_three_star_nominal = is_three_star & (mean_mag < dim_mag) & ~is_three_star_bright
        is_three_star_dim = is_three_star & (mean_mag >= dim_mag)

        mean_mag = np.mean([asterisms['star1_mag'], asterisms['star2_mag']], axis=0)
        is_two_star_bright = is_two_star & (mean_mag < nominal_mag)
        is_two_star_nominal = is_two_star & (mean_mag < dim_mag) & ~is_two_star_bright
        is_two_star_dim = is_two_star & (mean_mag >= dim_mag)

        description = "by Mean Mag"
    else:
        is_three_star_bright = is_three_star & (asterisms['star1_mag'] < nominal_mag) & (asterisms['star2_mag'] < nominal_mag) & (asterisms['star3_mag'] < nominal_mag)
        is_three_star_nominal = is_three_star & (asterisms['star1_mag'] < dim_mag) & (asterisms['star2_mag'] < dim_mag) & (asterisms['star3_mag'] < dim_mag) & ~is_three_star_bright
        is_three_star_dim = is_three_star & ((asterisms['star1_mag'] >= dim_mag) | (asterisms['star2_mag'] >= dim_mag) | (asterisms['star3_mag'] >= dim_mag))

        is_two_star_bright = is_two_star & (asterisms['star1_mag'] < nominal_mag) & (asterisms['star2_mag'] < nominal_mag)
        is_two_star_nominal = is_two_star & (asterisms['star1_mag'] < dim_mag) & (asterisms['star2_mag'] < dim_mag) & ~is_two_star_bright
        is_two_star_dim = is_two_star & ((asterisms['star1_mag'] >= dim_mag) | (asterisms['star2_mag'] >= dim_mag))

        description = "by All Mags"

    metrics = [
        ('SR', 'SR'),
        ('EE100', 'EE 100mas'),
        ('FWHM', 'FWHM'),
    ]

    colors = {
        'mean': {
            'bright': '#2ca02c',  # green
            'nominal': '#66c2a5',
            'dim': '#b2df8a',
        },
        'min': {
            'bright': '#1f77b4',  # blue
            'nominal': '#a6cee3',
            'dim': '#c6dbef',
        },
        'max': {
            'bright': '#d62728',  # red
            'nominal': '#fb9a99',
            'dim': '#fcae91',
        }
    }

    stats = ['min', 'mean', 'max']
    brightness_levels = ['nominal', 'dim', 'bright']

    for metric_key, metric_label in metrics:
        _, axs = plt.subplots(3, 2, figsize=(12, 10), sharex='row', sharey=True)
        for row, stat in enumerate(stats):
            for col, star_type in enumerate(['three', 'two']):
                ax = axs[row, col]
                for filt in brightness_levels:
                    mask = eval(f'is_{star_type}_star_{filt}')
                    ax.hist(
                        asterisms[f'{metric_key}_{stat}'][mask],
                        bins=20,
                        color=colors[stat][filt],
                        edgecolor='black',
                        alpha=0.8,
                        label=f'{filt.capitalize()}'
                    )
                ax.set_title(f"{stat.capitalize()} {'3★' if star_type == 'three' else '2★'}")
                ax.grid()
                if col == 0:
                    ax.set_ylabel("Count")
                if col == 1:
                    ax.legend(fontsize=8)

        plt.suptitle(f"{metric_label} {description} (H-band)", fontsize=14)
        plt.tight_layout()
        plt.show()
```

```python
plot_mag_comparison(asterisms, 'mean', nominal_mag=15.0, dim_mag=17.0)
```

```python
plot_mag_comparison(asterisms, "all", nominal_mag=16.0, dim_mag=18.0)
```
