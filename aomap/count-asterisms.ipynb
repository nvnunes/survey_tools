```python
from tabulate import tabulate
from astropy.coordinates import SkyCoord
from astropy.table import Table
import astropy.units as u
import numpy as np
```

```python
mag_min = 8.0
mag_max = 19.0
mag_step = 0.5

fits_path = '../output'
```

```python
# Load asterisms
fits_file = f"{fits_path}/asterisms-GNAO-Limit.fits"
asterisms = Table.read(fits_file, format='fits')
print('Total Asterisms:', len(asterisms))
print('Total 3-star Aserisms:', np.sum(asterisms['num_stars'] == 3))
```

```python
counts = Table(names=('mag_limit', 'count', 'change'), dtype=('f8', 'i4', 'i4'))
prev_count = 0
for mag in np.arange(mag_min, mag_max, mag_step):
    mask = (asterisms['star1_mag'] >= mag_min) & (asterisms['star1_mag'] < mag) & (asterisms['star2_mag'] >= mag_min) & (asterisms['star2_mag'] < mag) & (asterisms['star3_mag'] >= mag_min) & (asterisms['star3_mag'] <=mag)
    count = np.sum(mask)
    counts.add_row((mag, count, count - prev_count))
    prev_count = count
```

```python
display(tabulate(counts, headers=['R Limit', 'Count', 'Change'], floatfmt=['.1f', '.0f', '.0f'], tablefmt='html'))
```
